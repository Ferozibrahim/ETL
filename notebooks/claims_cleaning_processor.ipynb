```python
import pandas as pd
import numpy as np
df = pd.read_csv('../data/raw_claims.csv')
df.head(15)
```

```python
df['Claim_Date'] = pd.to_datetime(df['Claim_Date'], format='mixed')
df = df.dropna(subset=['Claim_Date'])
df.head(15)
```

```python
df['Claim_Amount'] = df['Claim_Amount'].replace('[\\$,]', '', regex=True).astype(float)
df = df.dropna(subset=['Claim_Amount']) 
df['Claim_Amount'] = df['Claim_Amount'].fillna(0)
df['Claim_Amount'] =df['Claim_Amount'].apply(lambda x: f"£{x:,.2f}")
df.head(15)
```

```python
df= df.dropna(subset=['Policy_Type'])
df = df.sort_values(by='Policy_Type', ascending=True)
df = df.reset_index(drop=True)
df.head(15)
```
